#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
# Calculate the mean, minimum, and maximum of each numeric variable
numeric_summary <- basketball_tidy |>
  summarise(
    across(
      where(is.numeric),
      list(
        mean = ~ mean(.x, na.rm = TRUE),
        min = ~ min(.x, na.rm = TRUE),
        max = ~ max(.x, na.rm = TRUE)
      )
    )
  )

# Display the summary table
numeric_summary
```

```r
# Plot the distribution of conference win percentages
ggplot(basketball_tidy, aes(x = conf_wins_pct)) +
  geom_histogram(
    binwidth = 5,
    fill = "steelblue",
    color = "white",
    na.rm = TRUE
  ) +
  labs(
    title = "Distribution of Conference Win Percentages",
    x = "Conference Win Percentage (%)",
    y = "Number of Tournament Appearances"
  ) +
  theme_minimal()
```

```r
# Calculate the proportion of each tournament finish category
finish_proportions <- basketball_tidy |>
  count(tourney_finish) |>
  mutate(proportion = n / sum(n))

# Plot the category proportions
ggplot(
  finish_proportions,
  aes(x = tourney_finish, y = proportion)
) +
  geom_col(fill = "steelblue") +
  coord_flip() +
  scale_x_discrete(labels = c(
    opening_round_loss = "Opening-round loss",
    first_round_loss = "First-round loss",
    second_round_loss = "Second-round loss",
    top_16_loss = "Round of 16 loss",
    top_8_loss = "Quarterfinal loss",
    top_4_loss = "Semifinal loss",
    top_2_loss = "Final loss",
    champ = "Champion"
  )) +
  scale_y_continuous(labels = scales::label_percent()) +
  labs(
    title = "Proportions of Tournament Finish Categories",
    x = "Tournament Finish",
    y = "Percentage of Tournament Appearances"
  ) +
  theme_minimal()
```

This plot compares seed distributions across tournament finish categories. It helps assess whether seed may be useful for predicting tournament finish. Differences between categories suggest predictive potential, while overlapping distributions indicate that seed alone may not distinguish all outcomes. Seed 0 is a special historical designation rather than a standard seed ranking.

```r
# Define categories from the earliest exit to champion
finish_order <- c(
  "opening_round_loss", "first_round_loss",
  "second_round_loss", "top_16_loss",
  "top_8_loss", "top_4_loss",
  "top_2_loss", "champ"
)

# Create readable category labels
finish_labels <- c(
  "Opening-round loss", "First-round loss",
  "Second-round loss", "Round of 16 loss",
  "Quarterfinal loss", "Semifinal loss",
  "Final loss", "Champion"
)

# Compare seed distributions across tournament outcomes
ggplot(basketball_tidy, aes(x = tourney_finish, y = seed)) +
  geom_boxplot(fill = "lightblue") +
  scale_x_discrete(
    limits = finish_order,
    labels = finish_labels
  ) +
  coord_flip() +
  labs(
    title = "Seed Distribution by Tournament Finish",
    x = "Tournament Finish",
    y = "Seed (0 = Special Historical Designation)"
  ) +
  theme_minimal()
```

This plot compares conference win percentage distributions across tournament finish categories, excluding missing percentages. It helps assess whether conference performance may be useful for predicting tournament finish alongside seed. Differences in medians and overlap between categories help evaluate how well this predictor might distinguish tournament outcomes.

```r
# Compare conference win percentages across tournament outcomes
basketball_tidy |>
  filter(!is.na(conf_wins_pct)) |>
  ggplot(aes(x = tourney_finish, y = conf_wins_pct)) +
  geom_boxplot(fill = "lightgreen") +
  scale_x_discrete(
    limits = finish_order,
    labels = finish_labels
  ) +
  coord_flip() +
  labs(
    title = "Conference Win Percentage by Tournament Finish",
    x = "Tournament Finish",
    y = "Conference Win Percentage (%)"
  ) +
  theme_minimal()
```
